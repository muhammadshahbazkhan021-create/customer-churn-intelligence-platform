```python
categorical_summary = df.describe(include=['object', 'category','str'])
print(categorical_summary)
```

```python
# 1. Get raw counts for each class
churn_counts = df['Churn'].value_counts()

# 2. Get percentages/ratios (multiply by 100 for percentages)
churn_percentages = df['Churn'].value_counts(normalize=True) * 100

# 3. Combine them into a clean, readable DataFrame
distribution_df = pd.DataFrame({
    'Count': churn_counts,
    'Percentage (%)': churn_percentages.round(2)
})

print("--- Churn Class Distribution ---")
print(distribution_df)
```
